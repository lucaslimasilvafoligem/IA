```python
def show_image(img):
    plt.imshow(img)
    plt.axis('off')
    plt.show()
```

```python
image2 = cv2.imread('../images/sovietic.jpg')
image2 = cv2.cvtColor(image2, cv2.COLOR_BGR2RGB)
show_image(image2)
```

```python
image2[100:200, 50:300] = np.array([0, 0, 0])
```

```python
show_image(image2)
```

```python
show_image(image2*image)
```

```python
show_image(cv2.add(image2, image))
```

```python
show_image(cv2.resize(image2, (100, 1000), fx=0.5, fy=0.5))
```

```python
show_image(cv2.addWeighted(image2, 0.5, image, 0.5, 0))
```

```python
#  Logo
logo = cv2.imread('../images/azul.jpg')
logo = cv2.cvtColor(logo, cv2.COLOR_BGR2RGB)
background = cv2.imread('../images/sovietic.jpg')
background = cv2.cvtColor(background, cv2.COLOR_BGR2RGB)
lh,lw = logo.shape[:2]
bh, bw = background.shape[:2]
ovr = np.zeros((bh, bw, 3), dtype='uint8')
ovr[0:lh, 0:lw] = logo
show_image(ovr)
```
