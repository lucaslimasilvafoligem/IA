```python
imagem_rgb = cv2.imread('../images/espace.jpg', cv2.IMREAD_COLOR_RGB)

# Upscale 2x com interpolação bicúbica
imagem_up = cv2.resize(imagem_rgb, None, fx=2, fy=2, interpolation=cv2.INTER_CUBIC)

# Remoção de ruído com filtro bilateral
imagem_denoise = cv2.bilateralFilter(imagem_up, d=9, sigmaColor=75, sigmaSpace=75)

# Aumento de nitidez com filtro de realce
kernel_nitidez = np.array([[0, -1, 0],
                           [-1, 5.5, -1],
                           [0, -1, 0]])
imagem_nitida = cv2.filter2D(imagem_denoise, -1, kernel_nitidez)

# Melhoria de contraste adaptativa
imagem_lab = cv2.cvtColor(imagem_nitida, cv2.COLOR_RGB2LAB)
l, a, b = cv2.split(imagem_lab)
clahe = cv2.createCLAHE(clipLimit=3.0, tileGridSize=(8, 8))
l_clahe = clahe.apply(l)
imagem_clahe = cv2.merge((l_clahe, a, b))
imagem_final = cv2.cvtColor(imagem_clahe, cv2.COLOR_LAB2RGB)

output_path = "../images/created/astronaut.jpg"
cv2.imwrite(output_path, cv2.cvtColor(imagem_final, cv2.COLOR_RGB2BGR))
```
