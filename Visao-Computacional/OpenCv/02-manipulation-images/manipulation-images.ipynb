```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
```

```python
image = Image.open('../images/azul.jpg')
type(image)
```

```python
image_vetor = np.asarray(image)
type(image_vetor)
```

```python
image_vetor.shape
```

```python
# Atualmente os pixels que formam as imagens possuem RGB
# Cada RGB possue cada uma das 3 cores principais cores
# Permitindo formar diferentes cores variando a intensidade das 3
# Vermelho, verde e azul
image_vetor
```

```python
plt.imshow(image_vetor[:,:,:])
```

```python
plt.imshow(image_vetor[:,:,0]) # Vermelho
```

```python
plt.imshow(image_vetor[:,:,1]) # Verde
```

```python
plt.imshow(image_vetor[:,:,2]) # Azul
```

```python
plt.imshow(image_vetor[:,:,2], cmap='gray')
```
