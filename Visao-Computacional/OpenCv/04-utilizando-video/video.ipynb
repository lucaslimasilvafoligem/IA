```python
import cv2
```

```python
video = cv2.VideoCapture(0)

while True:
    check, frame = video.read()
    
    if not check:
        break
    
    cv2.imshow("Video", frame)
    
    key = cv2.waitKey(1)
    if key == ord('q'):
        break
    
    cv2.putText(frame, "Pressione 'q' para sair", (10, 30), cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 255, 0), 2)
    cv2.imshow("Video", frame)
    
video.release()
cv2.destroyAllWindows()
```

```python
video = cv2.VideoCapture(0)
codec = cv2.VideoWriter_fourcc(*'XVID')
output = cv2.VideoWriter('output.avi', codec, 20.0, (640, 480))

while True:
    check, frame = video.read()

    if not check:
        break

    cv2.imshow("Video", frame)

    key = cv2.waitKey(1)
    if key == ord('q'):
        break

    cv2.putText(frame, "Pressione 'q' para sair", (10, 30), cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 255, 0), 2)
    cv2.imshow("Video", frame)

    output.write(frame)

output.release()
video.release()
cv2.destroyAllWindows()
```
